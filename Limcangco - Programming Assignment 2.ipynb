```python
import numpy as np   # Numpy is first imported as np in order for its functions to be used through the whole assignment.
```

#### A. Reproducible Normalization Problem

```python
np.random.seed(2112)                                # Sets random seed to 2112 for random values each time.
X = np.random.randint(10, 101, size=(5, 5))         # Creates a 5x5 array that generates random values from 10 - 100.
X
```

```python
X_mean = X.mean()                            # Calculates mean of all values in X.
X_std = X.std()                              # Calculates population standard deviation of all values in X.
X_normalized = (X - X_mean) / X_std          # Normalizes each value using mean and std.
X_normalized
```

```python
print('X:')                                           # Displays the original array.
print(X)
print('X_normalized:')                                # Displays the normalized array.
print(X_normalized)
print('Mean of X_normalized:')                        # Calculates the mean.
print(X_normalized.mean())
print('Standard deviation of X_normalized:')          # Calculates the population standard deviation.
print(X_normalized.std())
```

```python
np.save('X_normalized.npy', X_normalized)          # Saves the normalized array as the npy. file.
```

#### B. Cubes Divisible By 4 Problem

```python
C = np.arange(1, 101) ** 3          # Creates first 100 values and cubes.
C = C.reshape(10, 10)               # Reshapes cubed values into 10x10 array.
C
```

```python
div_by_4 = C[C % 4 == 0]          # Selects values divisible by 4.
div_by_4
```

```python
print('Shape of C:')                     # Displays dimensions for the array.
print(C.shape)
print('div_by_4:')                       # Displays values divisible by 4.
print(div_by_4)
print('Number of selected elements:')    # Displays the number of selected values.
print(div_by_4.size)
```

```python
np.save('div_by_4.npy', div_by_4)  # Saves the values as the npy. file.
```

#### C. Above-Mean Squares Problem

```python
S = np.arange(1, 37)**2  # Creates first 36 values and squares.
S = S.reshape(6, 6)      # Reshapes squared values into 6x6 array.
S
```

```python
S_mean = S.mean()        # Calculates mean of all values.
S_mean
```

```python
above_mean = S[S > S_mean]    # Selects values that are greater than the mean.
above_mean
```

```python
print('S:')                                     # Displays the array.
print(S)
print('S_mean:')                                # Displays the mean.
print(S_mean)
print('above_mean:')                            # Displays values greater than the mean.
print(above_mean)
print('Number of selected elements:')           # Displays number of selected values.
print(above_mean.size)
```

```python
np.save('above_mean.npy', above_mean)        # Saves selected values as the npy. file.
```
